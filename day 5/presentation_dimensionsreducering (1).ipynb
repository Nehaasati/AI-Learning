## Dimensionsreducering

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# Ladda data och dela in i X och y
data = load_iris()
X = data.data
y = data.target
```

```python
print("Antal dimensioner:", X.shape[1])

print("Features:", data.feature_names)
print("Target: ", data.target_names)
```

```python
# Standardisera (viktigt före PCA!)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# PCA
pca = PCA() 
X_pca = pca.fit_transform(X_scaled)
```

```python
print("Förklarad varians per komponent:", pca.explained_variance_ratio_)
print("Total förklarad varians:", sum(pca.explained_variance_ratio_))
```

```python
cumulative_variance = np.cumsum(pca.explained_variance_ratio_)
print(cumulative_variance)
```

```python
# PCA
pca = PCA(n_components=2) # Reducera till 2 komponenter
X_pca = pca.fit_transform(X_scaled)
```

```python
print("Förklarad varians per komponent:", pca.explained_variance_ratio_)
print("Total förklarad varians:", sum(pca.explained_variance_ratio_))
```

```python
# PCA
pca = PCA(n_components=0.95) # Reducera till så många komponenter som behövs för att behålla minst 95% av variansen
X_pca = pca.fit_transform(X_scaled)
```

```python
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA av Iris-datasetet")
plt.show()
```

Eftersom vi valde 2 dimensioner kan vi nu visualisera datan. 

PCA vet ingenting om klasserna. Det är en metod inom icke-väglett lärande. Den maximerar varians, inte klass-separation. 

Från grafen ser vi att en klass är tydligt separarad och att två klasser överlappar mer efter dimensionsreducering. De två överlappande klasserna kan alltså bli svåra att separera efter dimensionsreducering.
