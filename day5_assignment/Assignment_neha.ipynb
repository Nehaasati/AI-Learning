# 1. What is the curse of dimensionality?

The curse of dimensionality describes problems that appear when a dataset contains many features or dimensions. As dimensions increase, data points become sparse, making it harder for machine-learning algorithms to identify meaningful patterns. while the number of samples stays fixed. Distances between points start to look similar, making it harder for algorithms (especially distance-based ones like k-NN or k-means) to distinguish meaningful patterns. It also increases computational cost and the risk of overfitting, since models need exponentially more data to maintain the same sample density as dimensions grow. # 2. What is dimensionality reduction and why is it done?

Dimensionality reduction means reducing the number of features or variables used to represent data while keeping as much useful information as possible. It is performed to simplify datasets, reduce computational requirements, remove noise and redundancy, improve visualization, and sometimes reduce overfitting. A common method is Principal Component Analysis, or PCA, which transforms the original variables into new, uncorrelated components. The first components capture the greatest amount of variance. Instead of analyzing hundreds of original variables, a model might use a smaller number of components. This can make machine-learning algorithms faster and easier to apply, although some information may be lost

PCA, or Principal Component Analysis, transforms correlated original variables into new, uncorrelated variables called principal components. It first identifies directions in the data with the greatest variance. The first principal component captures the most variance, the second captures the next greatest amount while being perpendicular to the first, and so on. In Figure 5.4, the data points form an elongated pattern. Component C1 follows this main direction, so it preserves much of the variation. C2 is perpendicular and captures less variation. PCA then projects observations onto selected components, reducing dimensions while retaining most important information from the original dataset overall.
