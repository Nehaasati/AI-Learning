# 1. What is the curse of dimensionality?

The curse of dimensionality describes problems that appear when a dataset contains many features or dimensions. As dimensions increase, data points become sparse, making it harder for machine-learning algorithms to identify meaningful patterns. while the number of samples stays fixed. Distances between points start to look similar, making it harder for algorithms (especially distance-based ones like k-NN or k-means) to distinguish meaningful patterns. It also increases computational cost and the risk of overfitting, since models need exponentially more data to maintain the same sample density as dimensions grow. # 2. What is dimensionality reduction and why is it done?

Dimensionality reduction means reducing the number of features or variables used to represent data while keeping as much useful information as possible. It is performed to simplify datasets, reduce computational requirements, remove noise and redundancy, improve visualization, and sometimes reduce overfitting. A common method is Principal Component Analysis, or PCA, which transforms the original variables into new, uncorrelated components. The first components capture the greatest amount of variance. Instead of analyzing hundreds of original variables, a model might use a smaller number of components. This can make machine-learning algorithms faster and easier to apply, although some information may be lost

PCA (Principal Component Analysis) is a dimensionality-reduction method that transforms the original variables into new variables called principal components (PCs). The first principal component captures the largest possible amount of variation in the data, while the second captures the largest remaining variation and is perpendicular to the first. As shown in Figure 5.4, the data points can be projected onto a new axis, such as C1, while losing as little information as possible. PCA continues creating additional components until the desired amount of variance is explained. This allows many original variables to be represented using fewer dimensions

# 4. How can kernel PCA be evaluated?

Kernel PCA can be evaluated by measuring how well the transformed data performs in a machine-learning task, such as classification or regression. A pipeline can combine Kernel PCA with a predictive model, and GridSearchCV can then test different kernels and hyperparameters to find the best combination. For example, the RBF and sigmoid kernels can produce different representations of the data. The transformed data is evaluated based on the performance of the predictor, using metrics such as accuracy, precision, recall, F1-score, or mean squared error. The best Kernel PCA configuration is the one that provides the strongest validation performance.

# 6. After performing PCA, what happens to the interpretation of the variables?

After applying PCA, the interpretation of the original variables becomes more difficult. PCA creates new variables called principal components, which are combinations of the original variables. Each component can contain information from many original features. For example, PC1 might combine income, age, and experience into one new component. This makes it harder to say exactly what one variable contributes to the prediction. The advantage is that PCA can reduce the number of dimensions while preserving most of the important information. Therefore, PCA improves efficiency and can simplify datasets, but it usually reduces the interpretability and transparency of the original variables

The code demonstrates how PCA (Principal Component Analysis) can reduce the number of dimensions in a dataset.

import numpy as np
from sklearn.decomposition import PCA

First, NumPy and PCA from Scikit-learn are imported.

X = np.random.rand(1000, 3)

This creates a dataset with 1,000 observations and 3 features. The values are random numbers between 0 and 1.

pca = PCA(n_components=2)
X2D = pca.fit_transform(X)

Here, PCA is created to reduce the dataset from 3 dimensions to 2 dimensions. fit_transform() learns the most important directions in the data and transforms the original data into these two principal components.

X3D_inv = pca.inverse_transform(X2D)

This converts the 2-dimensional data back into 3 dimensions. However, the reconstructed data is only an approximation because some information was lost during dimensionality reduction.

print(np.allclose(X3D_inv, X))

This checks whether the reconstructed data is almost equal to the original data. It will normally return False, because reducing 3 dimensions to 2 causes some information to be lost
