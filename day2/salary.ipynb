```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection  import GridSearchCV, train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error

```

```python
df  = pd.read_csv("salary_dataset.csv")
```

```python
df.describe()
```

```python
df.head(6)
```

```python
x = df.drop(columns=['Salary'])
y = df['Salary']
```

```python
x_train_full, x_test_full, y_train_full, y_test_full = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
print(x_train_full.shape)
```

```python
print(x_train_full.head(6))
```

```python
from sklearn.model_selection import cross_validate


lr = LinearRegression()
linear_scores = cross_validate(
    lr,
    x_train_full,
    y_train_full,
    cv=5,
    scoring="neg_root_mean_squared_error"
)
```

```python
linear_rmse = -linear_scores["test_score"].mean()

print("\nLinear Regression")
print("RMSE:", linear_rmse)
```

```python


random_forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf_scores = cross_validate(
    random_forest_model,
    x_train_full,
    y_train_full,
    cv=5,
    scoring="neg_root_mean_squared_error"
)

rf_rmse = -rf_scores["test_score"].mean()

print("\nRandom Forest Regression")
print("RMSE:", rf_rmse)

```

```python
if linear_rmse < rf_rmse:
    best_model = lr
    print("\nBest model: Linear Regression")
else:
    best_model = random_forest_model
    print("\nBest model: Random Forest Regression")
```

```python

best_model.fit(x_train_full, y_train_full)

# Make predictions on the test set
y_pred = best_model.predict(x_test_full)
```

```python
# Evaluate the best model

from sklearn.metrics import r2_score


test_rmse = np.sqrt(mean_squared_error(y_test_full, y_pred))
test_r2 = r2_score(y_test_full, y_pred)

print("\nTest Set Evaluation")
print("RMSE:", test_rmse)
print("R²:", test_r2)


# Show actual vs predicted salaries

results = pd.DataFrame({
    "Actual Salary": y_test_full,
    "Predicted Salary": y_pred
})

print("\nActual vs Predicted:")
print(results)
```
