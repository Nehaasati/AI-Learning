# Chapter 6 – Clustering
# 1. What is clustering? Give some examples of application areas.

Clustering is an unsupervised machine learning technique used to divide data into groups called clusters. The observations in the same cluster are similar to each other, while observations in different clusters are more different. The computer finds the groups without already knowing the correct answers or labels. Clustering can be used in many areas. For example, companies can group customers according to their purchasing behaviour and preferences. It can also be used for market research, image recognition, recommendation systems, employee analysis, and geographical analysis. For example, customers with similar interests can receive more targeted advertisements.

# 2. Explain briefly how K-means works. Use Figures 6.3 and 6.4.

K-means is a clustering method that divides data into a chosen number of clusters. First, we choose the number of clusters, K. The algorithm places a centroid for each cluster. Every data point is assigned to the nearest centroid. Then the centroids are moved to the average position of the points belonging to each cluster. This process is repeated until the clusters stop changing significantly. Figure 6.3 shows different types of simulated clusters. Figure 6.4 shows that K-means works well when clusters are approximately spherical, similar in size, and have similar spread. It performs poorly when these conditions are not met.

# 3. How can we choose the number of clusters for K-means?

There is no single perfect method for choosing the number of clusters. One method is inertia, which measures how close the observations are to their cluster centroids. Lower inertia is better, but inertia always decreases when more clusters are added. Therefore, we can use the elbow method and look for a point where the improvement becomes much smaller. Another method is the silhouette score, which measures how well-separated and compact the clusters are. A higher score is generally better. A silhouette diagram shows the silhouette values for individual observations and helps us see whether the clusters are well separated and reasonably balanced.
