# 1. What characterizes classification problems? Give some examples of application areas.

A classification problem is a machine learning problem where we want to predict which category/class an observation belongs to.

The model receives some input features and predicts a class label.

For example, imagine we want to determine whether an email is spam:

Features: number of links, certain words, sender information, etc.
Output: Spam or Not spam

Here, the possible classes are discrete categories.

Classification vs. regression

It is important to distinguish classification from regression.

Classification:

Predicts a category.
Example: Is this transaction fraudulent or legitimate?
Example: Is a tumor benign or malignant?

Regression:

Predicts a numerical value.
Example: What will the house price be?
Example: What will tomorrow's temperature be?
Binary classification

If there are only two classes, it is called binary classification.

Examples:

Spam / Not spam
Fraud / Not fraud
Sick / Healthy
Pass / Fail
Multiclass classification

If there are more than two classes, it is called multiclass classification.

For example, suppose we want to classify an image as:

Cat
Dog
Horse
Bird

There are four possible classes.

Multilabel classification

There is also multilabel classification, where one observation can belong to several classes simultaneously.

For example, an image could contain:

Dog
Car
Tree

So the model can predict multiple labels at the same time.


# 2.Explain how OvR and OvO algorithms work.

OvR and OvO are methods for handling multiclass classification when the underlying classifier is primarily designed for binary classification.

OvR / One-vs-Rest

OvR means One-vs-Rest, also called One-vs-All.

Suppose we have three classes:

Cat
Dog
Horse

Instead of training one model to distinguish all three simultaneously, we train three binary classifiers.

Classifier 1

Cat vs. everything else:

Cat | Not Cat

Classifier 2

Dog vs. everything else:

Dog | Not Dog

Classifier 3

Horse vs. everything else:

Horse | Not Horse

When a new observation arrives, all three models produce a score/probability.

For example:

Classifier	Score
Cat vs Rest	0.20
Dog vs Rest	0.75
Horse vs Rest	0.35

The model chooses Dog, because it has the highest score.

Advantages of OvR
Relatively simple.
Requires only K classifiers for K classes.
Usually computationally cheaper than OvO when there are many classes.
Disadvantages

Each classifier must distinguish one class from all other classes, which can sometimes make the individual classification problems difficult or imbalanced.

OvO / One-vs-One

OvO means One-vs-One.

Instead of comparing each class against all the others, we create a classifier for every pair of classes.

For our three classes:

Cat vs Dog
Cat vs Horse
Dog vs Horse

That gives us:

{K(K-1)}{2} 

classifiers.

For 3 classes:

{3(3-1)}{2}=3 

For 10 classes:

{10(9)}{2}=45 

So the number of classifiers can become quite large.

Each classifier votes for one of its two classes.



# 3. Explain the following evaluation metrics
 # a) Confusion matrix

A confusion matrix shows how many predictions the classification model got right and wrong.

For binary classification, we normally have four important quantities:

	Actual Positive	Actual Negative
Predicted Positive	TP	FP
Predicted Negative	FN	TN
TP / True Positive

The model predicts positive, and the actual class is positive.

Example:

A patient has a disease, and the model predicts disease.

Correct prediction.

TN / True Negative

The model predicts negative, and the actual class is negative.

Example:

A patient does not have the disease, and the model predicts no disease.

Correct prediction.

FP / False Positive

The model predicts positive, but the actual class is negative.

Example:

The patient is healthy, but the model says they have the disease.

This is also called a false alarm.

FN / False Negative

The model predicts negative, but the actual class is positive.

Example:

The patient has the disease, but the model says they don't.

This can be particularly dangerous in medical applications.

# 3b. Accuracy

Accuracy tells us how many predictions were correct overall.

Accuracy={TP+TN}{TP+TN+FP+FN} 

For example, suppose we have 100 predictions:

TP = 40
TN = 50
FP = 5
FN = 5

Then:

Accuracy={40+50}{100}=90

So the model correctly classified 90% of the observations.

Problem with accuracy

Accuracy can be misleading when classes are imbalanced.

Imagine we have 1,000 transactions:

990 legitimate
10 fraudulent

A terrible model could simply predict:

"Every transaction is legitimate."

It would get:

{990}{1000}=99

accuracy.

That sounds excellent, but the model detects zero fraud cases.

Therefore, accuracy alone is often not enough.

# 3c. Precision

Precision answers:

"Of all the cases that the model predicted as positive, how many were actually positive?"

The formula is:

Precision={TP}{TP+FP} 

Suppose:

TP = 80
FP = 20

Then:

Precision={80}{80+20}=80

This means that when the model predicts positive, it is correct 80% of the time.

Example

Imagine a model identifies potentially fraudulent transactions.

The model flags 100 transactions as fraud.

80 really are fraud.
20 are legitimate.

Precision = 80%.

So precision is particularly important when false positives are costly.

# 3d. Recall

Recall is also called sensitivity or true positive rate.

Recall answers:

"Of all the actual positive cases, how many did the model successfully find?"

The formula is:

Recall={TP}{TP+FN} 

Suppose there are 100 actual positive cases.

The model correctly identifies 90 of them but misses 10.

Then:

Recall={90}{90+10}=90
Example

Imagine a medical model trying to detect a serious disease.

There are 100 people who actually have the disease.

The model identifies 95.

Then:

Recall=95

The model has missed 5 people.

A high recall means few positive cases are missed.

# 3e. F1-score

F1-score combines precision and recall into one metric.

It is the harmonic mean of precision and recall:

F1=2{Precision*Recall}{Precision+Recall} 

Why use the harmonic mean?

Because we want the model to perform well on both precision and recall.

For example:

Precision = 90%
Recall = 90%

Then:

 F1=90

But suppose:

Precision = 100%
Recall = 10%

The F1-score will be much lower.

This prevents a model from appearing good simply because one metric is extremely high while the other is terrible.

When is F1 useful?

F1-score is particularly useful when:

The classes are imbalanced.
Both false positives and false negatives matter.
We want a balance between precision and recall.
# 3f. ROC curve

ROC stands for Receiver Operating Characteristic.

The ROC curve evaluates how well a binary classifier separates two classes across different classification thresholds.

It plots:

True Positive Rate (TPR) on the y-axis
False Positive Rate (FPR) on the x-axis

Recall is the same as TPR:

TPR={TP}{TP+FN} 

False positive rate is:

$$ FPR={FP}{FP+TN}
What is a threshold?

Many classification models don't simply say:

Positive or negative.

Instead, they produce a probability or score.

For example:

Probability of fraud = 0.72

We might use a threshold of 0.5:

>= 0.5 - Positive
< 0.5 - Negative

But we could change the threshold to 0.8.

Now fewer observations will be classified as positive.

Changing the threshold changes both TPR and FPR.

The ROC curve shows this relationship.

AUC

We often summarize the ROC curve using AUC - Area Under the Curve.

AUC ranges from approximately:

1.0 - excellent separation
0.5 - roughly random guessing
below 0.5 - generally worse than random, assuming the direction is fixed

A higher AUC generally means the model is better at ranking positive cases above negative cases.




# 4. What is the precision-recall tradeoff?

The precision-recall tradeoff means that improving precision often comes at the cost of reducing recall, and vice versa.

This happens because we can change the classification threshold.

Imagine a model that predicts whether an email is spam.

Suppose the model gives every email a spam probability.

Low threshold

Suppose we classify an email as spam when:

Probability > 0.3

We will catch lots of spam.

Therefore:

Recall increases.

But we may also incorrectly classify legitimate emails as spam.

Therefore:

Precision decreases.

High threshold

Now suppose we use:

Probability > 0.9

We only classify emails as spam when we are very confident.

This reduces false positives.

Therefore:

Precision increases.

But we might miss many spam emails.

Therefore:

Recall decreases.

So there is often a tradeoff:

Higher precision ↔ lower recall

and

Higher recall ↔ lower precision

The best choice depends on the application.


# 5. Common classification models

# 5a. Logistic Regression

Despite its name, logistic regression is commonly used for classification.

It calculates a weighted combination of the input features and transforms the result into a probability.

For binary classification, imagine we have:

Age
Income
Number of previous purchases

The model calculates something like:

z=b_0+b_1x_1+b_2x_2+b_3x_3

It then applies the logistic/sigmoid function to convert this value into a probability between 0 and 1.

Conceptually:

 P(y=1)={1}{1+e^{-z}} 

For example:

Probability of customer leaving = 0.82

If the threshold is 0.5, the model predicts:

Customer will leave.

Advantages
Simple and fast.
Easy to interpret.
Works well when the relationship is approximately linear.
Produces probabilities.
Disadvantages
May not capture complex nonlinear relationships without additional feature engineering.
Can be affected by highly correlated features and feature scaling considerations.

# 5b. Support Vector Machines (SVM)

A Support Vector Machine tries to find a boundary that separates classes.

Imagine plotting two classes as points:

Red points
Blue points

The SVM tries to find a line that separates them.

The key idea is to find the boundary with the largest possible margin between the classes.

The observations closest to the decision boundary are called support vectors.

Why maximize the margin?

A larger margin can make the classifier more robust to new observations.

Nonlinear classification

SVMs can also handle nonlinear relationships using kernels.

For example, a kernel can allow the model to effectively work in a higher-dimensional feature space where the classes may become separable.

A commonly used kernel is the RBF (radial basis function) kernel.

Advantages
Powerful for high-dimensional data.
Can model nonlinear relationships using kernels.
Often works well when the number of features is large.
Disadvantages
Can be computationally expensive on large datasets.
Choice of parameters/kernel can be important.
Usually requires feature scaling for good performance.

# 5c. Decision Trees

A decision tree makes predictions by asking a sequence of questions about the features.

For example:

Is age > 30?
       /       \
     Yes        No
     /           \
Income > 50k?   Class A
   /    \
 Yes     No
 |        |
Class B  Class A

The tree starts at the root and repeatedly splits the data.

The goal is to create groups where the observations mostly belong to the same class.

Common measures for deciding how to split include:

Gini impurity
Entropy / information gain
Advantages
Easy to understand.
Easy to visualize.
Can model nonlinear relationships.
Does not require feature scaling.
Can work with different types of relationships.
Disadvantages

Decision trees can overfit.

For example, if we allow a tree to grow extremely deep, it might learn the training data almost perfectly but perform poorly on new data.

This is why parameters such as:

max_depth
min_samples_split
min_samples_leaf

can be important.

# 5d. Ensemble Learning

Ensemble learning means combining multiple models instead of relying on a single model.

The basic idea is:

Many models working together can produce a better prediction than one model alone.

There are several types of ensemble methods.

# VotingClassifier

A VotingClassifier combines different classifiers.

For example:

Logistic Regression
SVM
Decision Tree

Each model makes a prediction.

Suppose:

Logistic Regression - Cat
SVM -Dog
Decision Tree - Dog

The final prediction could be:

Dog

because Dog received the majority of votes.

This is called hard voting.

With soft voting, the models provide probabilities, and these probabilities are combined to determine the final prediction.

# BaggingClassifier

Bagging stands for Bootstrap Aggregating.

The idea is to train several models on different randomly sampled versions of the training data.

For example:

Training data
     ↓
 ┌───┼────┬────┐
 ↓   ↓    ↓    ↓
Model Model Model Model
  1    2    3    4
 └───┬────┴────┘
     ↓
 Combined prediction

The models are then combined.

Bagging can reduce variance and make predictions more stable.

# 5e. Random Forest

A Random Forest is an ensemble consisting of many decision trees.

Each tree is trained using a random sample of the training data, and the algorithm also introduces randomness when selecting features for splits.

For classification, the trees can vote.

Example:

Tree	Prediction
Tree 1	Dog
Tree 2	Cat
Tree 3	Dog
Tree 4	Dog
Tree 5	Cat

Votes:

Dog = 3
Cat = 2

Final prediction:

Dog

Why use many trees?

A single decision tree can be unstable and prone to overfitting.

Combining many different trees generally makes the model more robust.

Advantages
Usually strong predictive performance.
Handles nonlinear relationships.
Less prone to overfitting than a single unrestricted decision tree.
Can provide feature importance.
Usually does not require feature scaling.
Disadvantages
Less interpretable than a single decision tree.
Can require more memory and computation.
Feature importance can sometimes be misleading depending on the data and importance measure used.

# 5f. Extra Trees

Extra Trees, or Extremely Randomized Trees, is another tree-based ensemble method.

It is similar to Random Forest, but it introduces even more randomness when constructing the trees.

One important difference is how candidate split thresholds are selected.

Random Forest generally searches among candidate thresholds to find a good split.

Extra Trees introduces more randomization in choosing split thresholds and then selects among the resulting candidates.

The idea is that additional randomness can help create diverse trees.

Advantages
Strong performance on many classification problems.
Can handle nonlinear relationships.
Usually does not require feature scaling.
Can provide feature importance.
Randomization can reduce correlation between trees.
Disadvantages
Less interpretable than a single decision tree.
Can be computationally heavier than a single tree.
Feature importance still needs to be interpreted carefully.

# 6. What does feature importance mean in tree models?

Tree-based models can tell us which features were important for making predictions.

Suppose we are predicting whether someone will buy a product.

Our features could be:

Age
Income
Number of website visits
Previous purchases
Time spent on website

After training a Random Forest, we might get something like:

Feature	Importance
Previous purchases	0.40
Website visits	0.25
Income	0.20
Age	0.10
Time on website	0.05

This tells us that previous purchases contributed more to the model's predictions than age, according to the particular feature-importance method used.

How does a tree determine importance?

When a decision tree splits the data, it tries to make the resulting groups more pure.

If a feature repeatedly creates useful splits, that feature receives greater importance.

In an ensemble such as Random Forest, importance can be aggregated across many trees.

Important warning

Feature importance does not necessarily mean causality.

If the model says:

Feature X is very important.

That does not automatically mean:

Feature X causes the outcome.

It only means that the feature was useful for the model's predictions.

Also, correlated features can complicate interpretation. If two features contain similar information, the importance may be distributed between them or one may receive more importance than the other.

# 7. Discussion question — Precision vs. Recall

Stina says:

"I want the highest possible precision for our classification model."

Kalle responds:

"But what happens to recall?"

Kalle is pointing out the precision-recall tradeoff.

If we increase the classification threshold, the model becomes more conservative about predicting the positive class.

That can result in:

Higher precision

because the model only predicts positive when it is more confident.

But it can also result in:

Lower recall

because the model will miss more actual positive cases.

So you cannot simply say:

"The highest precision is always best."

It depends on the consequences of false positives and false negatives.

When do we want very high precision?

We want high precision when false positives are particularly costly or problematic.

Example: Spam detection

Suppose an email system is very aggressive about identifying spam.

If it incorrectly classifies an important work email as spam, the user might never see it.

Therefore, we may want high precision:

When the system says "spam," we want it to be very likely to actually be spam.

Example: Fraud investigation

Suppose a bank automatically freezes a customer's account when the system detects fraud.

A false positive could cause a legitimate customer's account to be frozen.

Therefore, high precision may be important.

The key principle

We need to consider the cost of different errors:

	Actual Positive	Actual Negative
Predicted Positive	True Positive	False Positive
Predicted Negative	False Negative	True Negative

The question is:

Which type of mistake is more harmful?

If false positives are especially harmful → we may prioritize precision.

If false negatives are especially harmful → we may prioritize recall.

# what is .fit_transform() vs .transform() is:First: What does a scaler do?

In the code from your page, you have something like:

X_train = scaler.fit_transform(X_train)

X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)

Here, scaler is probably something such as StandardScaler.

A scaler changes the numerical values of your features so that they are on a suitable scale.

For example, suppose we have:

Person	Age	Salary
A	20	30,000
B	30	50,000
C	40	70,000

Age is roughly between 20–40, while salary is between 30,000–70,000.

Some machine-learning models, especially Logistic Regression and SVM, can benefit from having features on similar scales.

A StandardScaler typically transforms a feature using:

$$ z = \frac{x-\mu}{\sigma} $$

where:

\(x\) = original value
\(\mu\) = mean
\(\sigma\) = standard deviation
\(z\) = scaled value
2. What does .fit() mean?

.fit() means:

Learn the parameters needed to perform the transformation.

For StandardScaler, .fit() calculates things such as:

the mean of each feature
the standard deviation of each feature

For example, imagine the training data contains:

Age:
20
30
40

The scaler calculates the mean:

$$ \mu = 30 $$

and the standard deviation.

It then remembers these values.

So conceptually:

scaler.fit(X_train)

means:

"Look at the training data and figure out how I should scale it."

3. What does .transform() mean?

.transform() means:

Use the parameters that were already learned to transform the data.

For example:

scaler.transform(X_train)

takes the mean and standard deviation that the scaler learned and uses them to scale the data.

Importantly, .transform() does not learn new parameters.

4. What does .fit_transform() mean?

.fit_transform() is basically two operations combined:

scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)

So:

X_train = scaler.fit_transform(X_train)

means:

Step 1

Learn the scaling parameters from X_train.

Step 2

Use those parameters to transform X_train.

5. Why do we use .fit_transform() on training data?

Because the training data is the data that we are allowed to learn from.

Suppose:

X_train

contains 80% of our dataset.

We want the scaler to learn:

"What does the training data look like?"

So we do:

scaler.fit_transform(X_train)

The scaler learns the mean and standard deviation from the training data and then scales it.

6. Why don't we use .fit_transform() on validation data?

This is the most important part.

Suppose we did:

X_val = scaler.fit_transform(X_val)

That would mean:

"Forget the scaling parameters from the training data and calculate new parameters using the validation data."

This is undesirable because the validation data is supposed to represent unseen data.

We don't want our preprocessing to learn anything from the validation set.

Instead, we do:

X_val = scaler.transform(X_val)

This means:

"Use the scaling rules that were learned from the training data and apply exactly the same rules to the validation data."

7. Why is this important?

Imagine we have:

Training data
Age: 20, 30, 40, 50

The scaler learns:

Mean = 35
Standard deviation = ...

Now suppose our validation data contains:

Age: 60, 70

We don't want to calculate a new mean using 60 and 70.

Instead, we use the mean and standard deviation from the training set.

So:

scaler.fit_transform(X_train)

followed by:

scaler.transform(X_val)

ensures that both datasets use the same transformation.

8. The same applies to the test data

The test data should represent completely unseen data.

Therefore:

X_test = scaler.transform(X_test)

We do not do:

X_test = scaler.fit_transform(X_test)

because that would allow information from the test data to influence the preprocessing.

The correct workflow is:

                 TRAINING DATA
                       |
                       ↓
                 scaler.fit()
                       |
              Learn mean/std
                       |
                       ↓
             scaler.transform()
                       |
                       ↓
                X_train_scaled


Validation data ──→ scaler.transform() ──→ X_val_scaled

Test data ────────→ scaler.transform() ──→ X_test_scaled

The important point is that the same scaler is used for all three datasets.

# 9. What is data leakage?

This is the big reason behind the rule.

Data leakage happens when information from data that should be unseen during training accidentally influences the model or preprocessing.

For example, imagine:

scaler.fit_transform(X_train)
scaler.fit_transform(X_test)

The second line calculates scaling parameters from the test data.

So the preprocessing has now "looked at" the test data.

Even though we haven't directly trained our model using y_test, we've allowed information from the test set to influence the preprocessing.

This can make our evaluation less trustworthy.
