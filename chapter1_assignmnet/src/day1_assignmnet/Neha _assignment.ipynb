# 1. How are AI, ML and DL connected?

Artificial Intelligence (AI) is the largest concept. It means creating computer systems that can perform tasks that normally require human intelligence, such as understanding language or recognizing objects. Machine Learning (ML) is a part of AI. ML allows computers to learn patterns from data instead of being explicitly programmed for every situation. Deep Learning (DL) is a part of ML. It uses artificial neural networks with many layers to learn complex patterns from large amounts of data. Therefore, the relationship can be described as: AI is the biggest area, ML is inside AI, and DL is inside ML.

# 2. What are the four problem categories in ML?

There are four important categories of machine-learning problems. The first is classification, where the model predicts a category, such as whether an email is spam or not spam. The second is regression, where the model predicts a numerical value, such as the price of a house. The third is clustering, where the model finds groups in data without already knowing the groups. For example, customers can be grouped according to their behavior. The fourth category is dimensionality reduction, where we reduce the number of features while keeping important information. These categories help us choose suitable machine-learning methods.

# 3a. Why divide data into training, validation and test data?

We divide data into training, validation, and test sets to evaluate a machine-learning model correctly. Training data is used to teach the model and learn its parameters. Validation data is used during development to compare models and choose hyperparameters. Test data is used only at the end to measure how well the final model works on unseen data. This separation is important because using the test data during development can cause overfitting. The model may become too specialized to the test data. A separate test set therefore gives us a more realistic estimate of performance on completely new data.

# 3b. What is k-fold cross-validation?

K-fold cross-validation is a method for evaluating a machine-learning model more reliably. First, the training data is divided into k equally sized parts, called folds. The model is trained using k-1 folds and tested on the remaining fold. This process is repeated k times, so every fold is used for validation once. Finally, the results are usually averaged. For example, with five-fold cross-validation, the data is divided into five parts and the model is trained and validated five times. Cross-validation is useful because it uses the available training data efficiently and gives a more stable estimate of model performance.

# 3c. What is Root Mean Squared Error (RMSE)?

Root Mean Squared Error, or RMSE, is a measurement used to evaluate regression models. It shows how far the model's predictions are from the actual values. First, we calculate the difference between each prediction and the real value. These differences are squared, then averaged. Finally, we take the square root. The formula is:

RMSE = √(average of squared errors)

A lower RMSE means better predictions because the predictions are closer to the actual values. RMSE uses the same unit as the target variable. For example, when predicting house prices in dollars, RMSE is also measured in dollars.

# 3d. What are hyperparameters and parameters?

A parameter is a value that a machine-learning model learns automatically from training data. For example, in linear regression, the coefficients are parameters because the model calculates them from the data. A hyperparameter is a value that we choose before or during model training. It controls how the model learns. For example, the alpha value in Lasso regression is a hyperparameter. Another example is the number of neighbors in K-nearest neighbors. Hyperparameters are normally selected using methods such as GridSearchCV. The important difference is that parameters are learned from data, while hyperparameters are selected by us.

# 3e. What are Grid Search and refit=True?

Grid Search is a method for finding good hyperparameter values. We create a grid containing different possible values, and the algorithm tests different combinations. For example, we might test Lasso with alpha values of 0.01, 0.1, 1, and 10. “Grid” refers to the combinations of possible values, while “search” means testing these combinations to find the best one. In scikit-learn, GridSearchCV also uses cross-validation. The parameter refit=True means that after finding the best hyperparameters, scikit-learn trains the best model again using all available training data.

# 3f. What is categorical data and how is it handled?

Categorical data contains values that represent groups or categories instead of numerical measurements. Nominal data has categories without an order, such as colors or countries. Ordinal data has categories with an order, such as small, medium, and large. One-hot encoding converts categories into separate columns containing zeros and ones. This is also called dummy-variable encoding when creating indicator variables for categories. Ordinal encoding converts ordered categories into numbers, such as small = 1, medium = 2, and large = 3. The correct encoding method depends on whether the categories have a meaningful order or not.

# 3g. What is feature engineering?

Feature engineering means creating, changing, or selecting input features to help a machine-learning model perform better. Features are the pieces of information that the model uses to make predictions. For example, if we have a date of birth, we could create a new feature called age. If we have house size and number of rooms, we could create a feature representing rooms per square meter. Feature engineering can also include removing useless features, combining features, or transforming values. Good features can make a model simpler and more accurate. Therefore, feature engineering is an important part of many machine-learning projects.

#3h. What is the principle of parsimony?

The principle of parsimony means that when several explanations or models can explain something, we should prefer the simplest one that works well. In machine learning, this means we should avoid making a model unnecessarily complicated. A very complicated model can learn noise and become overfitted. A simpler model is often easier to understand, faster to train, and easier to maintain. This does not mean that the simplest model is always best. Instead, we should choose a model that is complex enough to solve the problem but not more complex than necessary. This idea is also related to Occam's Razor.

# 4. What does “a model is a simplification of reality” mean?

A machine-learning model is a simplification of reality because real-world situations are extremely complicated. A model cannot normally include every possible factor that affects an outcome. Instead, it uses selected data and mathematical relationships to represent important patterns. For example, a house-price model may use location, size, income, and number of rooms, but it cannot include every detail about the house and its surroundings. The model therefore does not perfectly represent reality. Its purpose is to capture useful patterns that help us make predictions. A good model simplifies reality while still producing useful and reasonably accurate results.

# 5. What does it mean when a model is overfitted?

A model is overfitted when it learns the training data too closely, including random noise and unusual details that are not generally useful. An overfitted model can have excellent performance on training data but poor performance on new, unseen data. For example, a model might memorize specific examples instead of learning general patterns. Overfitting can happen when a model is too complicated compared with the amount of available data. We can reduce overfitting by using more training data, simpler models, regularization, cross-validation, or other techniques. The main goal of machine learning is to generalize well to new data.

# 6. What does “higher is better” mean in scikit-learn scoring?

In scikit-learn, the scoring system is designed so that higher scores always mean better performance. This can sometimes be confusing because some common error measurements, such as Mean Squared Error (MSE), are better when they are smaller. Scikit-learn solves this by using negative versions of these errors. For example, neg_mean_squared_error returns the negative MSE. A model with an MSE of 10 receives -10, while an MSE of 5 receives -5. Since -5 is higher than -10, scikit-learn correctly considers the model with MSE 5 to be better. Thus, “higher is better” is always maintained.

# 7. What are cross-sectional, time-series and panel data?

Cross-sectional data describes many subjects at one point in time. For example, information about 1,000 houses collected during the same year is cross-sectional data. Time-series data contains observations collected over time for one subject or system. For example, daily stock prices or monthly electricity usage are time-series data. Panel data combines both ideas. It follows multiple subjects over multiple points in time. For example, we could record the income of 500 people every year for ten years. Understanding the data type is important because different types require different analysis methods and can have different problems, especially regarding time.

# 9. Why does scikit-learn use neg_mean_squared_error?

Mean Squared Error (MSE) is normally an error, so a smaller value is better. However, scikit-learn follows the rule that higher scoring values are always better. To make MSE work with this rule, scikit-learn uses neg_mean_squared_error. It simply returns the negative value of MSE. For example, if a model has MSE = 100, the score becomes -100. If another model has MSE = 50, the score becomes -50. Because -50 is higher than -100, scikit-learn chooses the second model. Therefore, “negative” does not mean the model is worse. It is only a scoring convention.

# 11. Why do test_size=0.2 and then test_size=0.25 give 60-20-20?

First, test_size=0.2 puts 20% of the original data into the test set. Therefore, 80% remains as X_train_full. We then split this remaining 80% again using test_size=0.25. Twenty-five percent of the remaining 80% becomes validation data:

0.25 × 0.80 = 0.20 = 20%

The remaining 75% of the 80% becomes training data:

0.75 × 0.80 = 0.60 = 60%

Therefore, the final proportions are 60% training, 20% validation, and 20% test. This is why the second test_size is calculated from the remaining data, not from the original dataset.
