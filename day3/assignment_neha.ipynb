# Chapter 3 — Regression
1. What characterizes regression problems?


A regression problem is a machine learning problem where the goal is to predict a continuous numerical value.

For example, we might want to predict:

the price of a house
an employee's salary
the amount of electricity that will be consumed


The important difference between regression and classification is the type of output.

In classification, we predict a category, such as "yes/no", "spam/not spam", or "cat/dog".

In regression, we predict a number, such as €250,000, 72.5 kg, or 15.8 degrees.

MAE — Mean Absolute Error

MAE calculates the average absolute difference between the predicted values and the actual values.

Formula:

MAE = average(|actual value − predicted value|)

For example, if the errors are:

2, 3, 5

then:

MAE = (2 + 3 + 5) / 3 = 3.33

MAE is relatively easy to understand because it is expressed in the same unit as the target variable.

If we predict house prices in euros, the MAE is also in euros.

A disadvantage is that MAE does not punish very large errors as strongly as MSE or RMSE.

MSE — Mean Squared Error

MSE calculates the average of the squared errors.

MSE = average((actual value − predicted value)²)

Because the errors are squared, large errors become much more important.

For example, an error of 10 becomes:

10² = 100

while an error of 2 becomes:

2² = 4

Therefore, MSE is particularly sensitive to large errors and outliers.

The disadvantage is that the unit becomes squared. If the target is measured in euros, MSE is measured in euros², which is harder to interpret.

RMSE — Root Mean Squared Error

RMSE is simply the square root of MSE.

RMSE = √MSE

The important advantage is that RMSE has the same unit as the target variable.

Therefore, if we predict house prices in euros, RMSE is also expressed in euros.

Like MSE, RMSE gives relatively high importance to large errors.

# 6. Common regression models

a) Linear Regression

Linear regression tries to model the relationship between input variables and the target using a linear equation.

For one variable:

y = β₀ + β₁x

For several variables:

y = β₀ + β₁x₁ + β₂x₂ + ... + βₙxₙ

The model chooses coefficients that minimize the residual sum of squares, which corresponds to ordinary least squares.

For example, we could predict salary using:

years of experience
education
age

The advantage is that linear regression is relatively simple and easy 

Scikit-learn's LinearRegression implements ordinary 

b) Ridge Regression

Ridge regression is basically linear regression with L2 regularization.

Regularization adds a penalty for large coefficients.

This encourages the model to keep coefficients relatively small.

The idea is: alpha

"Don't make the coefficients unnecessarily large."

Ridge is useful when we have many features or when features are correlated with each other.

It can help reduce overfitting.

c) Lasso Regression

Lasso regression uses L1 regularization.

One important property of Lasso is that it can make some coefficients exactly zero.

That means Lasso can effectively perform feature selection.

For example, suppose we have 100 input variables.

Lasso might decide that some variables are not useful and give them coefficients of zero.

Therefore:

Lasso = regression + L1 regularization + possible feature selection.

Scikit-learn's Lasso minimizes a least-squares objective with an L1 penalty.

d) Elastic Net

Elastic Net combines Lasso and Ridge.

It uses both:

L1 regularization
L2 regularization

Therefore, Elastic Net can combine advantages of both approaches.

It can encourage some coefficients to become zero, like Lasso, while also shrinking coefficients, like Ridge.

This can be useful when there are many correlated features.

Scikit-learn describes Elastic Net as a linear regression model trained with both L1 and L2 regularization.

e) Support Vector Regression — SVR

Support Vector Regression is the regression version of Support Vector Machines.

Instead of  trying to minimize every error, SVR tries to find a function where many observations fall within a certain tolerance.

The tolerance is controlled by parameters such as epsilon.

SVR can also use kernels, which allow it to model nonlinear relationships.

For example, an RBF kernel can be used to capture nonlinear patterns.

SVR can work very well, but it can become computationally expensive with large datasets.

f) Decision Tree Regression

A decision tree divides the data into smaller groups using questions about the features.

For example:

Is age < 30?

If yes → go left.

If no → go right.

Then another question is asked.

Eventually, the observation reaches a leaf node, where the model produces a numerical prediction.

Decision trees can model nonlinear relationships and do not require the relationship between features and target to be linear.

They are also relatively easy to visualize and understand.

However, an unrestricted tree can easily overfit.

Scikit-learn's DecisionTreeRegressor supports parameters such as max_depth, min_samples_split, and min_samples_leaf that can be used to control tree complexity.

g) Ensemble Learning

Ensemble learning means combining several models instead of relying on only one model.



If the models make different errors, combining them can produce a better prediction.

Voting Regressor

A Voting Regressor fits several regression models and then combines their predictions, normally by taking their average.

For example:

Model 1 predicts: 100
Model 2 predicts: 110
Model 3 predicts: 105

The ensemble prediction could be:

(100 + 110 + 105) / 3 = 105

Scikit-learn's VotingRegressor combines predictions from several fitted regressors.

Bagging

Bagging also combines several models.

Different models are trained on different samples of the training data and their predictions are combined.

This can reduce variance and improve stability.

h) Random Forest

Random Forest is an ensemble of many decision trees.

Each tree is trained using a somewhat different sample of the data and a random subset of features.

The predictions from all trees are then combined, usually by averaging for regression.

For example:

Tree 1 → 100
Tree 2 → 105
Tree 3 → 98
Tree 4 → 110
Tree 5 → 102

The Random Forest prediction is approximately the average of these predictions.

Random Forest is powerful because many different trees are combined.

# 8. Difference between bagging and pasting

Both methods train several models using different subsets of the training data.

The difference is how the samples are created.

Bagging

Bagging = sampling with replacement.

This means an observation can appear more than once in a sample.

Example:

Original data:

A B C D E

One sample could be:

A C C D E

Here, C appears twice.

Pasting

Pasting = sampling without replacement.

An observation can only appear once in a particular sample.

Example:

A B C D E

One sample could be:

A C D E

but C cannot appear twice.

So the simple difference is:

Bagging → with replacement.

Pasting → without replacement.

 # 11. What is R²?

R² is called the coefficient of determination.

It measures how much of the variation in the target variable is explained by the model.

A common formula is:

R² = 1 − (SSres / SStot)

where:

SSres = residual sum of squares
SStot = total sum of squares

A simple way to understand it is:

R² of:

1.0 → perfect predictions.

0.0 → equivalent to predicting the mean in the standard case.

Less than 0 → the model can be worse
